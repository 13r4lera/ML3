1. Построение простого графика

Напишите код, который строит график функции у = $x^2$ на интервале [-10, 10].
Добавьте заголовок, подписи осей и сетку.

```python
import numpy as np
import matplotlib.pyplot as plt


x = np.linspace(-10, 10, 100)
y = x**2
plt.plot(x, y)
plt.title('График функции y = x²')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

2. Настройка стилей и цветов

Постройте три линии на одном графике:

 у = х (синяя, пунктирная линия),  
 у = $x^2$ (зеленая, штрих-пунктирная линия),  
 у = $x^3$ (красная, сплошная линия).  
Добавьте легенду и сделайте оси одинакового масштаба.

```python
x = np.linspace(-2, 2, 200)
y1 = x
y2 = x**2
y3 = x**3

plt.plot(x, y1, 'b--', label='y = x')      
plt.plot(x, y2, 'g-.', label='y = x²')     
plt.plot(x, y3, 'r-', label='y = x³')      
plt.title('Графики функций')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.axis('equal')

plt.show()
```

7. Трехмерный график поверхности

Используя mpl_toolkits. mplot3d , постройте 3D-график функции z = sin(vx2+y2)
на сетке значений х, у в диапазоне [-5,5].
    

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)

z = np.sin(x**2 + y**2)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(x, y, z, cmap='viridis')

ax.set_title('z = sin(x² + y²)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

fig.colorbar(surf, ax=ax)

plt.show()
```

8. Множественные подграфики (subplots)

Постройте четыре графика в одной фигуре (2 х 2 сетка):

1. Линейный график у = х  
2. Парабола у = $x^2$  
3. Синус у = sin(x)  
4. Косинус у = cos(x)  
Добавьте заголовки к каждому подграфику.

```python
x = np.linspace(-10, 10, 400)

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

axs[0, 0].plot(x, x)
axs[0, 0].set_title('y = x')
axs[0, 0].grid(True)

axs[0, 1].plot(x, x**2)
axs[0, 1].set_title('y = x²')
axs[0, 1].grid(True)

axs[1, 0].plot(x, np.sin(x))
axs[1, 0].set_title('y = sin(x)')
axs[1, 0].grid(True)

axs[1, 1].plot(x, np.cos(x))
axs[1, 1].set_title('y = cos(x)')
axs[1, 1].grid(True)

fig.suptitle('Множественные подграфики', fontsize=14)
plt.show()
```

Вариант 12. Площадь под модулем синуса   
Определите интеграл:   
f(x) =|sin(x)|   
на отрезке [0, 2π].

```python
f = lambda x: np.abs(np.sin(x))

a, b = 0, 2*np.pi

x = np.linspace(a, b, 1000)
y = f(x)

area = np.trapezoid(y, x)

plt.plot(x, y, label='|sin(x)|', color='blue')

plt.fill_between(x, y, color='blue', alpha=0.3)

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

plt.title('Площадь под |sin(x)| на [0, 2π]')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()

plt.show()

print("Площадь под кривой:", area)
```

12. Двухгорбая поверхность (бигауссовский холм)  
Постройте поверхность:  
$f(x,y) = e^{-((x-1)2+y2)} + e^{-((z+1)2+y2)}$  
на х, у є [-3, 3].


```python
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
x, y = np.meshgrid(x, y)

z = np.exp(-((x - 1)**2 + (y - 1)**2)) + np.exp(-((x + 1)**2 + (y + 1)**2))

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(x, y, z, cmap='viridis', alpha=0.8)

ax.plot_wireframe(x, y, z, color='black', linewidth=0.5)

ax.set_title('Двухгорбовая поверхность (surface + wireframe)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

fig.colorbar(surf, ax=ax, shrink=0.6)

plt.show()
```
